Creating Tensor CPU example

```python
import torch
first_tens = torch.tensor([[12,10,11,9],[13,15,14,16]])
second_tens = torch.tensor([[1,2,3,4],[5,6,7,8]])
add_tens = first_tens + second_tens
print(add_tens)
print(add_tens.size())
```

```python
sub_tens = first_tens - second_tens
print(sub_tens)
print(sub_tens.size())
```

Creating tensor GPU example

```python
import torch
print(torch.__version__)
```

```python
if torch.cuda.is_available(): device = "cuda"
else : device = "cpu"
print(device)
```

```python
tens_a = torch.tensor([[10,11,12,13],[14,15,16,17]], device = device)
tens_b = torch.tensor([[18,19,20,21],[22,23,24,25]], device = device)
multi_tens = tens_a * tens_b
print(multi_tens)
```

Different ways to create tensors

```python
import torch
import numpy as np

tensor_from_list = torch.tensor([[1,2,3,4],[5,6,7,8]])
tensor_from_tuple = torch.tensor(((1,2,3,4),(5,6,7,8)))
print("tensor from list")
print(tensor_from_list)
print("tensor from tuple")
print(tensor_from_tuple)
```

```python
tensor_from_array = torch.tensor(np.array([11,12,13,14,15]))
print(tensor_from_array)
```

Different functions for creating tensors


```python
tensor_emp = torch.empty(3,4)
print("empty tensor", tensor_emp)
```

```python
tensor_zeros = torch.zeros(3,4)
print("zeros tensor", tensor_zeros)
```

```python
tensor_ones = torch.ones(3,4)
print("ones tensor", tensor_ones)
```

```python
#returns a tensor filled with random numbers from a uniform distribution
tensor_rand_un = torch.rand(3,4)
print("random tensor", tensor_rand_un)
tensor_rand_norm = torch.randn(3,4)
print("random normal tensor", tensor_rand_norm)
```
